### 实现模型预测函数

线性回归模型: $f_{w,b}(x) = wx + b$

```python
# 设置模型参数
w = 100
b = 100
print(f"w: {w}, b: {b}")
```

```python
def compute_model_output(x, w, b):
    """
    计算线性回归模型的预测值
    
    参数:
        x (ndarray): 输入特征
        w (scalar): 权重
        b (scalar): 偏置
    
    返回:
        f_wb (ndarray): 预测值
    """
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

```python
# 计算预测值
tmp_f_wb = compute_model_output(x_train, w, b)
print(f"预测值: {tmp_f_wb}")
```

### 实现代价函数

代价函数 (均方误差): $J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} (f_{w,b}(x^{(i)}) - y^{(i)})^2$

```python
def compute_cost(x, y, w, b):
    """
    计算线性回归的代价函数
    
    参数:
        x (ndarray): 输入特征
        y (ndarray): 目标值
        w (scalar): 权重
        b (scalar): 偏置
    
    返回:
        total_cost (float): 代价
    """
    m = x.shape[0]
    
    cost_sum = 0
    for i in range(m):
        f_wb = w * x[i] + b
        cost = (f_wb - y[i]) ** 2
        cost_sum += cost
    total_cost = (1 / (2 * m)) * cost_sum
    
    return total_cost
```

## 练习 3: 梯度下降 (Gradient Descent)

实现梯度下降算法来优化模型参数

```python
import math, copy
from lab_utils_uni import plt_house_x, plt_contour_wgrad, plt_divergence, plt_gradients
```

```python
# 重新定义训练数据
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
```

### 实现梯度计算函数

梯度公式:
- $\frac{\partial J(w,b)}{\partial w} = \frac{1}{m} \sum_{i=1}^{m} (f_{w,b}(x^{(i)}) - y^{(i)})x^{(i)}$
- $\frac{\partial J(w,b)}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} (f_{w,b}(x^{(i)}) - y^{(i)})$

```python
def compute_gradient(x, y, w, b):
    """
    计算梯度
    
    参数:
        x (ndarray): 输入特征
        y (ndarray): 目标值
        w (scalar): 权重
        b (scalar): 偏置
    
    返回:
        dj_dw (scalar): 代价函数对w的梯度
        dj_db (scalar): 代价函数对b的梯度
    """
    m = x.shape[0]
    dj_dw = 0
    dj_db = 0
    
    for i in range(m):
        f_wb = w * x[i] + b
        dj_dw_i = (f_wb - y[i]) * x[i]
        dj_db_i = f_wb - y[i]
        dj_db += dj_db_i
        dj_dw += dj_dw_i
    dj_dw = 1 / m * dj_dw
    dj_db = 1 / m * dj_db
    
    return dj_dw, dj_db
```

```python
# 可视化梯度
plt_gradients(x_train, y_train, compute_cost, compute_gradient)
plt.show()
```

### 实现梯度下降算法

更新规则:
- $w = w - \alpha \frac{\partial J(w,b)}{\partial w}$
- $b = b - \alpha \frac{\partial J(w,b)}{\partial b}$

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iter, cost_function, gradient_function):
    """
    执行梯度下降以优化w和b
    
    参数:
        x (ndarray): 输入特征
        y (ndarray): 目标值
        w_in (scalar): 初始权重
        b_in (scalar): 初始偏置
        alpha (float): 学习率
        num_iter (int): 迭代次数
        cost_function: 代价函数
        gradient_function: 梯度函数
    
    返回:
        w (scalar): 优化后的权重
        b (scalar): 优化后的偏置
        J_history (list): 代价历史
        p_history (list): 参数历史
    """
    w = copy.deepcopy(w_in)
    J_history = []
    p_history = []
    b = b_in
    w = w_in
    
    for i in range(num_iter):
        # 计算梯度
        dj_dw, dj_db = gradient_function(x, y, w, b)
        
        # 更新参数
        b = b - alpha * dj_db
        w = w - alpha * dj_dw
        
        # 保存历史
        if i < 100000:
            J_history.append(cost_function(x, y, w, b))
            p_history.append([w, b])
        
        # 打印进度
        if i % math.ceil(num_iter / 10) == 0:
            print(f'迭代次数 {i:4}: 损失函数 {J_history[-1]:0.2e}',
                  f"dj_dw:{dj_dw: 0.3e}, dj_db:{dj_db: 0.3e}",
                  f"w:{w: 0.3e}, b:{b: 0.5e}"
                  )
    
    return w, b, J_history, p_history
```

### 运行梯度下降

```python
# 初始化参数
w_in = 0
b_in = 0
iterations = 40000
tmp_alpha = 1.0e-3

# 运行梯度下降
w_final, b_final, J_hist, p_hist = gradient_descent(
    x_train, y_train, w_in, b_in, tmp_alpha, iterations, compute_cost, compute_gradient
)

print(f'\n最终的w: {w_final:0.2f}, 最终的b: {b_final:0.2f}')
```

### 可视化训练过程

```python
# 绘制代价函数随迭代次数的变化
fig, (ax1, ax2) = plt.subplots(1, 2, constrained_layout=True, figsize=(10, 5))
ax1.plot(J_hist[:1000])
ax2.plot(1000 + np.arange(len(J_hist[1000:])), J_hist[1000:])
ax1.set_title("Cost vs. iteration(start)")
ax2.set_title("Cost vs. iteration (end)")
ax1.set_ylabel('Cost')
ax2.set_ylabel('Cost')
ax1.set_xlabel('iteration step')
ax2.set_xlabel('iteration step')
plt.show()
```
